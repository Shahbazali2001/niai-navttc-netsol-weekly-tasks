## 3. Python Fundamentals: Tasks 3 to 10
Hands-on implementations of arithmetic operations, type conversions, complex numbers, geometric calculations, and formula conversions.


```python
# ==========================================
# Task 3: Python Program to Add Two Numbers
# ==========================================
num1 = 1.5
num2 = 6.3

# Add two numbers
sum_val = num1 + num2

# Display the sum
print("--- Task 3: Addition ---")
print("The sum of {0} and {1} is {2}".format(num1, num2, sum_val))
```

```python
# ==========================================
# Task 4: Add Two Numbers with User Input
# ==========================================
# In automated testing/Colab, we provide mock input or default fallback
def task_4_add_inputs(val1=12.5, val2=7.5):
    print("--- Task 4: User Input Addition ---")
    # For interactive usage:
    # num1 = input('Enter first number: ')
    # num2 = input('Enter second number: ')
    num1 = str(val1)
    num2 = str(val2)
    sum_result = float(num1) + float(num2)
    print("The sum of {0} and {1} is {2}".format(num1, num2, sum_result))
    return sum_result

task_4_add_inputs()
```

```python
# ==========================================
# Task 5: Calculate Square Root of a Number
# ==========================================
num = 8

# Square root calculation using exponentiation
num_sqrt = num ** 0.5
print("--- Task 5: Square Root ---")
print("The square root of %0.3f is %0.3f" % (num, num_sqrt))

# Test with perfect squares
for val in [16, 25, 49, 144]:
    print(f"Square root of {val} is {val ** 0.5:.2f}")
```

```python
# ==========================================
# Task 6: Square Root of Real or Complex Numbers
# ==========================================
import cmath

print("--- Task 6: Complex Square Root ---")
num_complex = 1 + 2j

# Compute complex square root
num_sqrt_complex = cmath.sqrt(num_complex)
print("The square root of {0} is {1:0.3f}+{2:0.3f}j".format(
    num_complex, num_sqrt_complex.real, num_sqrt_complex.imag
))

# Negative real number yields imaginary unit
neg_num = -16
print(f"The square root of {neg_num} is {cmath.sqrt(neg_num)}")
```

```python
# ==========================================
# Task 7: Celsius to Fahrenheit Conversion
# ==========================================
celsius = 37.5

# Formula: F = (C * 1.8) + 32
fahrenheit = (celsius * 1.8) + 32

print("--- Task 7: Temperature Conversion ---")
print("%0.1f degree Celsius is equal to %0.1f degree Fahrenheit" % (celsius, fahrenheit))

# Conversion back: C = (F - 32) / 1.8
c_back = (fahrenheit - 32) / 1.8
print(f"Verification: {fahrenheit:.1f}°F converts back to {c_back:.1f}°C")
```

```python
# ==========================================
# Task 8: Area of Triangle (Heron's Formula)
# ==========================================
a = 5
b = 6
c = 7

# Calculate semi-perimeter
s = (a + b + c) / 2

# Calculate area using Heron's formula
area = (s * (s - a) * (s - b) * (s - c)) ** 0.5

print("--- Task 8: Area of Triangle ---")
print("Sides: a=%0.1f, b=%0.1f, c=%0.1f" % (a, b, c))
print("Semi-perimeter (s): %0.2f" % s)
print("The area of the triangle is %0.2f" % area)
```

```python
# ==========================================
# Task 9: Solve Quadratic Equation ax^2 + bx + c = 0
# ==========================================
import cmath

print("--- Task 9: Quadratic Equation Solver ---")
# Coefficients: a*x^2 + b*x + c = 0
a = 1
b = 5
c = 6

# Calculate discriminant: d = b^2 - 4ac
d = (b ** 2) - (4 * a * c)

# Compute roots using quadratic formula
sol1 = (-b - cmath.sqrt(d)) / (2 * a)
sol2 = (-b + cmath.sqrt(d)) / (2 * a)

print("Coefficients: a={0}, b={1}, c={2}".format(a, b, c))
print("Discriminant: {0}".format(d))
print("The solutions are {0} and {1}".format(sol1, sol2))

# Example with complex roots: x^2 + 2x + 5 = 0
a_c, b_c, c_c = 1, 2, 5
d_c = (b_c**2) - (4 * a_c * c_c)
root1 = (-b_c + cmath.sqrt(d_c)) / (2 * a_c)
root2 = (-b_c - cmath.sqrt(d_c)) / (2 * a_c)
print(f"Complex roots of x^2 + 2x + 5 = 0: {root1} and {root2}")
```

```python
# ==========================================
# Task 10: Convert Kilometers to Miles
# ==========================================
kilometers = 100.0

# Conversion factor
conv_fac = 0.621371

# Calculate miles
miles = kilometers * conv_fac

print("--- Task 10: Kilometers to Miles ---")
print(f"{kilometers} kilometers is equal to {miles:.4f} miles")

# Additional examples:
distances_km = [5, 10, 21.097, 42.195]  # Marathon distances
for km in distances_km:
    print(f"  {km:6.2f} km = {km * conv_fac:6.2f} miles")
```

## 4. Conditional Statements: Tasks 11 to 14
Execution of `if`, `if-else`, nested conditionals, and `if-elif-else` ladders.


```python
# ==========================================
# Task 11: Python if Statement
# ==========================================
print("--- Task 11: if Statement ---")
i = 10
if (i > 15):
    print("10 is less than 15")
print("I am Not in if")

# Let's test with a value where the condition evaluates to True
i = 20
if (i > 15):
    print("20 is greater than 15 (Inside if block)")
```

```python
# ==========================================
# Task 12: Python if-else Statement
# ==========================================
print("--- Task 12: if-else Statement ---")
i = 20
if (i < 15):
    print("i is smaller than 15")
    print("i'm in if Block")
else:
    print("i is greater than 15")
    print("i'm in else Block")
print("i'm not in if and not in else Block")
```

```python
# ==========================================
# Task 13: Python Nested if Statement
# ==========================================
print("--- Task 13: Nested if Statement ---")
i = 10
if (i == 10):
    # First if statement
    if (i < 15):
        print("i is smaller than 15")
    # Nested - if statement
    # Will only be executed if statement above is true
    if (i < 12):
        print("i is smaller than 12 too")
    else:
        print("i is greater than 15")
```

```python
# ==========================================
# Task 14: Python if-elif-else Ladder
# ==========================================
print("--- Task 14: if-elif-else Ladder ---")
i = 20
if (i == 10):
    print("i is 10")
elif (i == 15):
    print("i is 15")
elif (i == 20):
    print("i is 20")
else:
    print("i is not present")

# Practical Grading System example with elif
score = 87
if score >= 90:
    grade = "A+"
elif score >= 80:
    grade = "A"
elif score >= 70:
    grade = "B"
else:
    grade = "C"
print(f"Score: {score} -> Grade: {grade}")
```

## 5. Loops and Iterations: Tasks 15 to 19
Mastering `for` loops, `while` loops, loop termination with `break`, loop stepping with `continue`, and advanced looping techniques (`enumerate`, `zip`, `reversed`, `sorted`).


```python
# ==========================================
# Task 15: Exercises on for Loops in Python
# ==========================================
print("--- Task 15: for Loops ---")

# Iterating over range
print("Range iteration 1 to 5:")
for val in range(1, 6):
    print(val, end=" ")
print()

# Iterating over a list
fruits = ["apple", "banana", "cherry", "date"]
print("\nIterating over items:")
for fruit in fruits:
    print(f"Fruit: {fruit}")

# Iterating with step
print("\nEven numbers up to 10:")
for num in range(2, 11, 2):
    print(num, end=" ")
print()
```

```python
# ==========================================
# Task 16: Exercises on while Loops in Python
# ==========================================
print("--- Task 16: while Loops ---")

# Countdown counter
count = 5
while count > 0:
    print(f"Countdown: {count}")
    count -= 1
print("Blastoff!")

# Accumulating sum until threshold
total = 0
i = 1
while total < 20:
    total += i
    print(f"Added {i}, current sum = {total}")
    i += 1
```

```python
# ==========================================
# Task 17: Exercises on break Statement
# ==========================================
print("--- Task 17: break Statement ---")

# Terminating loop when search target is found
numbers = [10, 25, 42, 73, 99, 105]
target = 42

for index, num in enumerate(numbers):
    if num == target:
        print(f"Found target {target} at index {index}!")
        break
    print(f"Checking index {index}: {num}")
```

```python
# ==========================================
# Task 18: Exercises on continue Statement
# ==========================================
print("--- Task 18: continue Statement ---")

# Skipping odd numbers to process only evens
print("Filtering evens using continue:")
for num in range(1, 11):
    if num % 2 != 0:
        continue  # skip rest of loop body for odd numbers
    print(f"Processed even number: {num}")
```

```python
# ==========================================
# Task 19: Various Looping Techniques
# ==========================================
print("--- Task 19: Advanced Looping Techniques ---")

# 1. Using enumerate() to track indices
languages = ["Python", "C++", "Java", "Julia"]
print("1. enumerate():")
for idx, lang in enumerate(languages, start=1):
    print(f"  Rank #{idx}: {lang}")

# 2. Using zip() to iterate parallel sequences
students = ["Ali", "Sara", "Zain"]
scores = [92, 98, 85]
grades = ["A+", "A+", "A"]
print("\n2. zip():")
for student, score, grade in zip(students, scores, grades):
    print(f"  {student:<6}: Score={score}, Grade={grade}")

# 3. Using reversed()
print("\n3. reversed():")
for lang in reversed(languages):
    print(f"  {lang}", end=" ")
print()

# 4. Using sorted() without modifying original
unsorted_nums = [45, 12, 89, 3, 67]
print("\n4. sorted():")
for n in sorted(unsorted_nums):
    print(n, end=" ")
print()

# 5. Using dict.items()
course_meta = {"name": "AI/ML NAVTTC", "duration_weeks": 12, "level": "Advanced"}
print("\n5. dict.items():")
for k, v in course_meta.items():
    print(f"  {k} -> {v}")
```

## 6. Core Python Data Structures: Tasks 21 to 25
Comprehensive exercises covering Lists, Tuples, Strings, Sets, and Dictionaries.


```python
# ==========================================
# Task 21: Exercise on List Data Type
# ==========================================
print("--- Task 21: List Operations ---")
my_list = [10, 20, 30, 40, 50]
print(f"Initial list: {my_list}")

# List methods
my_list.append(60)              # append to end
my_list.insert(2, 25)           # insert at index 2
my_list.extend([70, 80])        # extend with iterable
print(f"After additions: {my_list}")

popped_item = my_list.pop()    # remove and return last item
my_list.remove(25)              # remove first occurrence of 25
print(f"Popped: {popped_item}, after removal: {my_list}")

# Slicing
print(f"Sub-slice [1:4]: {my_list[1:4]}")
print(f"Reversed slice [::-1]: {my_list[::-1]}")

# List Comprehension
squares = [x**2 for x in range(1, 8)]
print(f"List comprehension (squares): {squares}")
```

```python
# ==========================================
# Task 22: Exercise on Tuple Data Type
# ==========================================
print("--- Task 22: Tuple Operations ---")
# Tuples are ordered, immutable collections
point = (4, 5, 9)
single_tuple = (42,)  # Note the trailing comma

print(f"3D Point: {point}, type: {type(point)}")
print(f"Point x={point[0]}, y={point[1]}, z={point[2]}")

# Tuple Unpacking
x, y, z = point
print(f"Unpacked variables: x={x}, y={y}, z={z}")

# Immutability Demonstration
try:
    point[0] = 10
except TypeError as e:
    print(f"Caught expected immutability error: {e}")

# Tuple count and index methods
data_tuple = (1, 2, 2, 3, 2, 4)
print(f"Count of '2': {data_tuple.count(2)}")
print(f"Index of '3': {data_tuple.index(3)}")
```

```python
# ==========================================
# Task 23: Exercise on String Data Type
# ==========================================
print("--- Task 23: String Operations ---")
sample_text = "  Artificial Intelligence and Machine Learning  "
print(f"Original: '{sample_text}'")

# String manipulation methods
cleaned_text = sample_text.strip()
print(f"Stripped: '{cleaned_text}'")
print(f"Uppercase: {cleaned_text.upper()}")
print(f"Lowercase: {cleaned_text.lower()}")
print(f"Replaced: {cleaned_text.replace('Machine', 'Deep')}")

# Splitting and Joining
words = cleaned_text.split(" ")
print(f"Tokens/Words: {words}")
hyphenated = "-".join([w for w in words if w])
print(f"Hyphenated: {hyphenated}")

# String Slicing & Formatting
greeting = "Hello"
target = "Colab User"
f_string = f"{greeting}, {target}! String length = {len(cleaned_text)}"
print(f"Formatted: {f_string}")
```

```python
# ==========================================
# Task 24: Exercise on Set Data Type
# ==========================================
print("--- Task 24: Set Operations ---")
# Sets store unique, unordered, unindexed elements
set_a = {1, 2, 3, 4, 5, 5, 2}
set_b = {4, 5, 6, 7, 8}

print(f"Set A (duplicates removed): {set_a}")
print(f"Set B: {set_b}")

# Mathematical Set Operations
print(f"Union (A | B): {set_a | set_b}")
print(f"Intersection (A & B): {set_a & set_b}")
print(f"Difference (A - B): {set_a - set_b}")
print(f"Symmetric Difference (A ^ B): {set_a ^ set_b}")

# Set methods
set_a.add(10)
set_a.discard(1)  # discards without error if not present
print(f"Modified Set A: {set_a}")
```

```python
# ==========================================
# Task 25: Exercise on Dictionary Data Type
# ==========================================
print("--- Task 25: Dictionary Operations ---")
# Dictionaries store key-value pairs
student_record = {
    "id": 101,
    "name": "Hamza Tariq",
    "course": "AI NAVTTC",
    "skills": ["Python", "Linux", "Git"],
    "enrolled": True
}

print(f"Student Record: {student_record}")
print(f"Name: {student_record['name']}")
print(f"Non-existent key with default: {student_record.get('gpa', 'N/A')}")

# Modifying and adding entries
student_record["gpa"] = 3.85
student_record["skills"].append("Machine Learning")

print(f"Updated Record: {student_record}")
print(f"Keys: {list(student_record.keys())}")
print(f"Values: {list(student_record.values())}")

# Dictionary Comprehension
squared_dict = {x: x**2 for x in range(1, 6)}
print(f"Dictionary comprehension: {squared_dict}")
```
