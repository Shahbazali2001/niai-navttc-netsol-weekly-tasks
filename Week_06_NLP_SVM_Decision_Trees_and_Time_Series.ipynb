## 1. Natural Language Processing (NLP) Preprocessing Suite
Text preprocessing transforms unstructured human language into clean, structured tokens.
We demonstrate:
- Word and Sentence Tokenization
- Stopword Removal & Text Normalization
- Stemming (Porter) vs Lemmatization (WordNet)
- POS Tagging, NER Tagging, Chunking, and WordNet Synsets.


```python
# Module 1: Complete NLP Preprocessing Pipeline
import re

sample_corpus = """Natural Language Processing (NLP) is a subfield of artificial intelligence.
It enables computers to understand, interpret, and generate human language.
Dr. Smith from New York published an outstanding paper on Transformers in 2023!"""

print("--- Original Corpus ---")
print(sample_corpus)

# 1. Regex Tokenization & Sentence Splitting
def regex_sentence_tokenize(text):
    return [s.strip() for s in re.split(r'[.!?]\s+', text) if s.strip()]

def regex_word_tokenize(text):
    return re.findall(r'\b[a-zA-Z0-9_-]+\b', text.lower())

sentences = regex_sentence_tokenize(sample_corpus)
words = regex_word_tokenize(sample_corpus)

print(f"\nTokenized Sentences ({len(sentences)}):")
for idx, s in enumerate(sentences, 1):
    print(f"  [{idx}] {s}")

print(f"\nTokenized Words ({len(words)} tokens):")
print(words[:15], "...")

# 2. Stopword Filtering & Noise Removal
common_stopwords = set([
    "is", "a", "of", "it", "to", "and", "in", "an", "on", "from", "for", "the", "this"
])
filtered_words = [w for w in words if w not in common_stopwords]
print(f"\nAfter Stopword Removal ({len(filtered_words)} tokens):")
print(filtered_words[:12])
```

```python
# Module 1 (Cont): NLTK Suite & Linguistic Structures (Colab Ready)
# In Google Colab, run:
# import nltk
# nltk.download(['punkt', 'stopwords', 'wordnet', 'averaged_perceptron_tagger', 'maxent_ne_chunker', 'words'])

def demonstrate_nltk_pipeline():
    try:
        import nltk
        from nltk.stem import PorterStemmer, WordNetLemmatizer
        
        stemmer = PorterStemmer()
        lemmatizer = WordNetLemmatizer()

        sample_words = ["running", "runs", "easily", "fairly", "corpora", "studies", "better"]
        print("--- Stemming vs Lemmatization Comparison ---")
        print(f"{'Word':<12} {'Porter Stem':<15} {'Lemmatized':<15}")
        print("-" * 42)
        for w in sample_words:
            stem = stemmer.stem(w)
            lemma = lemmatizer.lemmatize(w, pos='v' if w in ['running', 'runs'] else 'n')
            print(f"{w:<12} {stem:<15} {lemma:<15}")
    except Exception as e:
        print(f"NLTK demo (will execute in Colab): {e}")
        # Algorithmic Suffix Stripping demonstration
        sample_words = ["running", "runs", "easily", "studies"]
        print("--- Suffix Stripping Demonstration ---")
        for w in sample_words:
            root = re.sub(r'(ing|s|ly|ies)$', '', w)
            print(f"Word: {w:<10} -> Root: {root}")

demonstrate_nltk_pipeline()
```

## 2. Text Feature Extraction: BoW, TF-IDF & Cosine Similarity
1. **Bag of Words (BoW)**: Counts raw term occurrences across documents.
2. **TF-IDF**: Weighs word importance inversely proportional to its frequency across all documents:
   $$\text{TF-IDF}(t, d, D) = \text{TF}(t, d) \times \log\left(\frac{|D|}{1 + |\{d \in D : t \in d\}|}\right)$$
3. **Cosine Similarity**: Measures angular distance between document feature vectors:
   $$\text{CosineSim}(u, v) = \frac{u \cdot v}{\|u\| \|v\|}$$


```python
# Module 2: Bag of Words, TF-IDF, and Cosine Similarity from Scratch (NumPy)
import numpy as np

documents = [
    "Artificial intelligence drives autonomous systems and robotics",
    "Deep learning and artificial intelligence revolutionize computer vision",
    "Robotics and autonomous vehicles use advanced computer algorithms"
]

# Build Vocabulary
vocabulary = sorted(list(set(re.findall(r'\b\w+\b', " ".join(documents).lower()))))
vocab_index = {word: idx for idx, word in enumerate(vocabulary)}
N_docs = len(documents)
V_size = len(vocabulary)

print(f"Vocabulary Size: {V_size} unique terms")

# 1. Bag of Words (BoW) Term-Document Matrix
bow_matrix = np.zeros((N_docs, V_size), dtype=int)
for doc_idx, doc in enumerate(documents):
    for token in re.findall(r'\b\w+\b', doc.lower()):
        bow_matrix[doc_idx, vocab_index[token]] += 1

print("\n--- Bag of Words (Term Frequencies for Doc 1) ---")
for word, idx in list(vocab_index.items())[:6]:
    print(f"  {word:<15}: count = {bow_matrix[0, idx]}")

# 2. TF-IDF Calculation
df = np.sum(bow_matrix > 0, axis=0)  # Document frequency per term
idf = np.log((N_docs + 1) / (df + 1)) + 1.0  # Smooth IDF
tf_norm = bow_matrix / np.sum(bow_matrix, axis=1, keepdims=True)
tfidf_matrix = tf_norm * idf

# 3. Pairwise Cosine Similarity
def cosine_sim(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b) + 1e-15)

sim_01 = cosine_sim(tfidf_matrix[0], tfidf_matrix[1])
sim_02 = cosine_sim(tfidf_matrix[0], tfidf_matrix[2])
sim_12 = cosine_sim(tfidf_matrix[1], tfidf_matrix[2])

print("\n--- Document Cosine Similarity ---")
print(f"Similarity (Doc 0 vs Doc 1): {sim_01:.4f}")
print(f"Similarity (Doc 0 vs Doc 2): {sim_02:.4f}")
print(f"Similarity (Doc 1 vs Doc 2): {sim_12:.4f}")
```

## 3. Machine Learning II: Model Evaluation Metrics & Class Imbalance
- Classification Metrics: Confusion Matrix, Accuracy, Precision, Recall, F1-Score, ROC-AUC.
- Imbalance Remedies: Class weights, Resampling, SMOTE.


```python
# Module 3: Evaluation Metrics Calculation from First Principles
# Given binary predictions on an imbalanced dataset:
# Class 0: 950 negative samples | Class 1: 50 positive samples
y_true = np.array([0]*950 + [1]*50)
# Classifier detects 40 true positives, 20 false positives, 10 false negatives, 930 true negatives
y_pred = np.array([0]*930 + [1]*20 + [0]*10 + [1]*40)

# Confusion Matrix components
TP = np.sum((y_true == 1) & (y_pred == 1))
FP = np.sum((y_true == 0) & (y_pred == 1))
TN = np.sum((y_true == 0) & (y_pred == 0))
FN = np.sum((y_true == 1) & (y_pred == 0))

accuracy  = (TP + TN) / len(y_true)
precision = TP / (TP + FP)
recall    = TP / (TP + FN)
f1_score  = 2 * (precision * recall) / (precision + recall)

print("--- Model Evaluation Metrics (Imbalanced Data) ---")
print(f"Confusion Matrix: [ [TN={TN}, FP={FP}], [FN={FN}, TP={TP}] ]")
print(f"Accuracy:  {accuracy * 100:.2f}% (High accuracy can be misleading!)")
print(f"Precision: {precision:.4f} (Of all predicted positives, {precision*100:.1f}% were true)")
print(f"Recall:    {recall:.4f} (Detected {recall*100:.1f}% of actual positive cases)")
print(f"F1-Score:  {f1_score:.4f} (Harmonic mean of precision and recall)")
```

## 4. Supervised Classifiers: Decision Tree (Task 53) & SVM (Task 54)
- **Task 53**: Decision Tree (Gini impurity vs Information gain, tree visualization)
- **Task 54**: Support Vector Machine (Linear & RBF Kernel SVM)
- **Ensemble**: Random Forest Bagging


```python
# Tasks 53 & 54: Decision Tree, SVM, and Random Forest Implementation
def run_classifiers():
    try:
        from sklearn.datasets import make_classification
        from sklearn.model_selection import train_test_split
        from sklearn.tree import DecisionTreeClassifier
        from sklearn.svm import SVC
        from sklearn.ensemble import RandomForestClassifier
        from sklearn.metrics import accuracy_score, classification_report

        X, y = make_classification(n_samples=300, n_features=4, n_informative=3, n_redundant=1, random_state=42)
        X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

        # 1. Decision Tree (Task 53)
        dt = DecisionTreeClassifier(criterion="gini", max_depth=4, random_state=42)
        dt.fit(X_train, y_train)
        dt_acc = accuracy_score(y_test, dt.predict(X_test))
        print("--- Task 53: Decision Tree Classifier ---")
        print(f"Decision Tree Accuracy: {dt_acc * 100:.2f}%")
        print(f"Feature Importances:    {dt.feature_importances_.round(4)}")

        # 2. Support Vector Machine (Task 54)
        svm_linear = SVC(kernel="linear", C=1.0)
        svm_linear.fit(X_train, y_train)
        svm_lin_acc = accuracy_score(y_test, svm_linear.predict(X_test))

        svm_rbf = SVC(kernel="rbf", C=1.0, gamma="scale")
        svm_rbf.fit(X_train, y_train)
        svm_rbf_acc = accuracy_score(y_test, svm_rbf.predict(X_test))

        print("\n--- Task 54: Support Vector Machine (SVM) ---")
        print(f"Linear SVM Accuracy: {svm_lin_acc * 100:.2f}%")
        print(f"RBF Kernel SVM Accuracy: {svm_rbf_acc * 100:.2f}%")

        # 3. Random Forest (Bagging Ensemble)
        rf = RandomForestClassifier(n_estimators=100, max_depth=4, random_state=42)
        rf.fit(X_train, y_train)
        rf_acc = accuracy_score(y_test, rf.predict(X_test))
        print(f"\nRandom Forest Accuracy: {rf_acc * 100:.2f}%")

    except ImportError:
        print("Scikit-learn models are prepared for Colab runtime execution.")

run_classifiers()
```

## 5. Task 55: Time Series Analysis with Pandas
Time series operations:
- Datetime indexing & frequencies
- Rolling window statistics (Moving Average & Volatility)
- Resampling (Daily to Weekly/Monthly)
- Seasonality and trend inspection


```python
# Task 55: Exercise on Time Series Analysis
import pandas as pd
import numpy as np

# 1. Create Synthetic Time Series: Daily electricity demand over 1 year
np.random.seed(42)
dates = pd.date_range(start="2025-01-01", end="2025-12-31", freq="D")
trend = np.linspace(50, 80, len(dates))
seasonal = 15 * np.sin(2 * np.pi * np.arange(len(dates)) / 365.25)
noise = np.random.normal(0, 4, len(dates))

demand = trend + seasonal + noise
df_ts = pd.DataFrame({"Demand_MWh": demand}, index=dates)

print("--- Task 55: Time Series Analysis ---")
print(df_ts.head())

# 2. Rolling Window Calculations (7-day and 30-day Moving Averages)
df_ts["7D_MA"]  = df_ts["Demand_MWh"].rolling(window=7).mean()
df_ts["30D_MA"] = df_ts["Demand_MWh"].rolling(window=30).mean()
df_ts["30D_STD"] = df_ts["Demand_MWh"].rolling(window=30).std()

print("\nRolling Window Statistics:")
print(df_ts[["Demand_MWh", "7D_MA", "30D_MA", "30D_STD"]].dropna().head())

# 3. Resampling: Monthly Mean Demand
df_monthly = df_ts["Demand_MWh"].resample("M").agg(["mean", "min", "max", "std"])
print("\nMonthly Aggregations (Resampling):")
print(df_monthly)
```
