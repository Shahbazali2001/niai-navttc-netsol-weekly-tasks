```python
# ==========================================
# Task 43: Exercise on Pandas Series Data
# ==========================================
print("--- Task 43: Pandas Series ---")
# Creating Series from python list with custom index labels
temperatures = [28.5, 30.2, 33.0, 31.5, 29.8]
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
temp_series = pd.Series(temperatures, index=days, name="Daily_Temp_Celsius")

print(f"Temperature Series:\n{temp_series}")
print(f"\nTemperature on Wednesday: {temp_series['Wed']}°C")
print(f"Mean temperature: {temp_series.mean():.2f}°C")
print(f"Max temperature:  {temp_series.max():.2f}°C")
print(f"Days above 30°C:\n{temp_series[temp_series > 30.0]}")
```

```python
# ==========================================
# Task 44: Exercise on Pandas DataFrame
# ==========================================
print("--- Task 44: Pandas DataFrame ---")
# Creating DataFrame from dictionary of lists
data_dict = {
    "StudentID": [101, 102, 103, 104, 105],
    "Name": ["Ali", "Sara", "Zain", "Fatima", "Bilal"],
    "Department": ["AI", "CS", "AI", "SE", "CS"],
    "CGPA": [3.8, 3.9, 3.4, 3.7, 3.2],
    "Attendance": [95, 98, 82, 91, 78]
}

df_students = pd.DataFrame(data_dict)
print(f"Students DataFrame:\n{df_students}")

# Indexing: .loc (label-based) and .iloc (position-based)
print(f"\nRow 0 using .iloc:\n{df_students.iloc[0]}")
print(f"\nSubset (Rows 1-3, Columns Name & CGPA):\n{df_students.loc[1:3, ['Name', 'CGPA']]}")

# Conditional Selection
high_achievers = df_students[(df_students["CGPA"] >= 3.5) & (df_students["Attendance"] > 90)]
print(f"\nHigh Achievers (CGPA >= 3.5 and Attendance > 90%):\n{high_achievers}")
```

```python
# ==========================================
# Task 45: Exercise on Pandas Open CSV Files
# ==========================================
print("--- Task 45: Reading and Writing CSV Files ---")
# 1. Export DataFrame to CSV
csv_filename = "student_data.csv"
df_students.to_csv(csv_filename, index=False)
print(f"Saved DataFrame to disk as '{csv_filename}'")

# 2. Ingest CSV file back into Pandas
df_loaded = pd.read_csv(csv_filename)
print(f"Loaded DataFrame from CSV successfully:")
print(df_loaded.head(3))
```

```python
# ==========================================
# Task 46: Exercise on Pandas Data Analyzation
# ==========================================
print("--- Task 46: Data Analysis Methods ---")
print("1. Shape of dataset:", df_students.shape)
print("\n2. Summary Info:")
df_students.info()

print("\n3. Descriptive Statistics:")
print(df_students.describe())

print("\n4. Unique departments:", df_students["Department"].unique())
print("5. Department value counts:")
print(df_students["Department"].value_counts())

print("\n6. Sorting by CGPA (Descending):")
print(df_students.sort_values(by="CGPA", ascending=False))
```

```python
# ==========================================
# Task 47: Data Cleaning Techniques
# ==========================================
print("--- Task 47: Data Cleaning ---")
# Create messy dataset with missing values (NaN) and duplicates
dirty_data = {
    "Name": ["A", "B", "C", "D", "E", "A"],
    "Age": [22, np.nan, 25, 23, np.nan, 22],
    "Salary": [50000, 60000, np.nan, 75000, 52000, 50000]
}
df_dirty = pd.DataFrame(dirty_data)
print(f"Dirty DataFrame with nulls and duplicate:\n{df_dirty}")

# 1. Check for Missing Values
print(f"\nNull counts per column:\n{df_dirty.isnull().sum()}")

# 2. Remove Duplicates
df_cleaned = df_dirty.drop_duplicates()
print(f"\nAfter dropping duplicates (Rows {len(df_dirty)} -> {len(df_cleaned)}):\n{df_cleaned}")

# 3. Impute Missing Values (Age with median, Salary with mean)
df_cleaned = df_cleaned.copy()
df_cleaned["Age"] = df_cleaned["Age"].fillna(df_cleaned["Age"].median())
df_cleaned["Salary"] = df_cleaned["Salary"].fillna(df_cleaned["Salary"].mean())

print(f"\nAfter Imputing Missing Values:\n{df_cleaned}")
```

```python
# ==========================================
# Task 48: Exercise on Pandas Data Correlation
# ==========================================
print("--- Task 48: Correlation Analysis ---")
# Calculating correlation matrix across numeric features
corr_matrix = df_students[["CGPA", "Attendance"]].corr()
print(f"Correlation Matrix (CGPA vs Attendance):\n{corr_matrix}")
print(f"Pearson r value: {corr_matrix.loc['CGPA', 'Attendance']:.4f}")
```

## 2. Advanced Pandas Operations
- Merging, Joining, and Concatenating datasets
- GroupBy split-apply-combine workflow
- Discretization & Binning (`pd.cut`, `pd.qcut`)


```python
# Advanced Pandas: Merging, GroupBy, and Binning
# 1. Joins: Merging two DataFrames on key
df_dept_info = pd.DataFrame({
    "Department": ["AI", "CS", "SE", "EE"],
    "HeadOfDept": ["Dr. Usman", "Dr. Ayesha", "Dr. Farhan", "Dr. Kamran"]
})

merged_df = pd.merge(df_students, df_dept_info, on="Department", how="left")
print("--- Merged DataFrame (Left Join) ---")
print(merged_df[["StudentID", "Name", "Department", "HeadOfDept", "CGPA"]])

# 2. GroupBy Aggregation
dept_summary = df_students.groupby("Department").agg(
    Avg_CGPA=("CGPA", "mean"),
    Avg_Attendance=("Attendance", "mean"),
    Student_Count=("StudentID", "count")
)
print(f"\n--- GroupBy Department Summary ---\n{dept_summary}")

# 3. Discretization & Binning: Categorizing CGPA into performance tiers
df_students["Tier"] = pd.cut(
    df_students["CGPA"],
    bins=[0, 3.3, 3.7, 4.0],
    labels=["Bronze", "Silver", "Gold"]
)
print(f"\n--- Binning Results ---\n{df_students[['Name', 'CGPA', 'Tier']]}")
```

## 4. Supervised Machine Learning: Univariate Linear Regression with Gradient Descent
### Mathematical Formulation
1. **Hypothesis**:
   $$h_\theta(x) = \theta_0 + \theta_1 x$$
2. **Mean Squared Error (MSE) Cost Function**:
   $$J(\theta_0, \theta_1) = \frac{1}{2m} \sum_{i=1}^m (h_\theta(x^{(i)}) - y^{(i)})^2$$
3. **Gradient Descent Update Rules**:
   $$\theta_0 := \theta_0 - \alpha \frac{1}{m} \sum_{i=1}^m (h_\theta(x^{(i)}) - y^{(i)})$$
   $$\theta_1 := \theta_1 - \alpha \frac{1}{m} \sum_{i=1}^m (h_\theta(x^{(i)}) - y^{(i)}) x^{(i)}$$

Below are the two requested implementations:
- **Without Vectorization** (Iterative loop updates)
- **With Vectorization** (NumPy vectorized dot products)


```python
# Implementation 1: Univariate Linear Regression WITHOUT Vectorization (Pure Loops)
# Generate synthetic 1D regression dataset: y = 2.5 * x + 4.0 + noise
np.random.seed(42)
m = 100
X_raw = np.random.uniform(1.0, 10.0, m)
y_raw = 2.5 * X_raw + 4.0 + np.random.normal(0, 1.2, m)

def gradient_descent_unvectorized(X, y, alpha=0.01, epochs=500):
    m = len(X)
    theta0 = 0.0
    theta1 = 0.0
    cost_history = []

    for epoch in range(epochs):
        # Calculate cost J(theta) iteratively
        total_error = 0.0
        grad0 = 0.0
        grad1 = 0.0

        for i in range(m):
            prediction = theta0 + theta1 * X[i]
            error = prediction - y[i]
            total_error += error ** 2
            grad0 += error
            grad1 += error * X[i]

        cost = total_error / (2 * m)
        cost_history.append(cost)

        # Simultaneous parameter update
        theta0 -= alpha * (grad0 / m)
        theta1 -= alpha * (grad1 / m)

        if epoch % 100 == 0:
            print(f"Epoch {epoch:3d} | Cost: {cost:.4f} | theta0: {theta0:.4f}, theta1: {theta1:.4f}")

    return theta0, theta1, cost_history

print("--- Running Unvectorized Gradient Descent ---")
th0_unvec, th1_unvec, costs_unvec = gradient_descent_unvectorized(X_raw, y_raw, alpha=0.02, epochs=500)
print(f"\nFinal Parameters (Unvectorized): theta0 = {th0_unvec:.4f}, theta1 = {th1_unvec:.4f}")
```

```python
# Implementation 2: Univariate Linear Regression WITH Vectorization (NumPy Matrix Operations)
def gradient_descent_vectorized(X, y, alpha=0.02, epochs=500):
    m = len(y)
    # Add intercept column of ones to X: X_b shape (m, 2)
    X_b = np.c_[np.ones(m), X]
    theta = np.zeros(2)  # [theta0, theta1]
    cost_history = []

    for epoch in range(epochs):
        # Vectorized hypothesis prediction: h = X_b . theta
        predictions = X_b.dot(theta)
        errors = predictions - y

        # Vectorized MSE cost
        cost = (1 / (2 * m)) * np.sum(errors ** 2)
        cost_history.append(cost)

        # Vectorized gradient calculation: (1/m) * X_b^T . errors
        gradients = (1 / m) * X_b.T.dot(errors)

        # Parameter update
        theta -= alpha * gradients

    return theta, cost_history

print("--- Running Vectorized Gradient Descent ---")
theta_vec, costs_vec = gradient_descent_vectorized(X_raw, y_raw, alpha=0.02, epochs=500)
print(f"Final Parameters (Vectorized): theta0 = {theta_vec[0]:.4f}, theta1 = {theta_vec[1]:.4f}")
print(f"Difference between methods: |theta0_diff| = {abs(th0_unvec - theta_vec[0]):.6f}, |theta1_diff| = {abs(th1_unvec - theta_vec[1]):.6f}")

# Plot Cost Convergence & Fitted Line
plt.figure(figsize=(12, 4))

# Plot 1: Cost history
plt.subplot(1, 2, 1)
plt.plot(costs_vec, color="firebrick", lw=2)
plt.title("MSE Cost vs Training Epochs")
plt.xlabel("Epoch")
plt.ylabel("Cost J(theta)")

# Plot 2: Fitted regression line
plt.subplot(1, 2, 2)
plt.scatter(X_raw, y_raw, color="steelblue", alpha=0.7, label="Observed Data")
x_line = np.linspace(1, 10, 100)
y_line = theta_vec[0] + theta_vec[1] * x_line
plt.plot(x_line, y_line, color="red", lw=2, label=f"Fit: y = {theta_vec[1]:.2f}x + {theta_vec[0]:.2f}")
plt.title("Univariate Linear Regression Fit")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.tight_layout()
plt.savefig("gradient_descent_fit.png", bbox_inches="tight")
plt.close()
print("Saved gradient_descent_fit.png")
```
