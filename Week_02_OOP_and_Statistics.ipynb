## 1. Task 20: Advanced Functions, Lambdas, and Scope
Covers function definitions, keyword parameters, variable positional and keyword arguments, local vs global scope, lambda expressions, higher-order functions (`map`, `filter`), and nested functions/closures.


```python
# Task 20: Functions, Variable Scope, and Advanced Arguments
# 1. Basic Function and Default Parameters
def calculate_cylinder_volume(radius, height=10.0):
    """Calculates volume of a cylinder with default height."""
    import math
    return math.pi * (radius ** 2) * height

print(f"Volume (radius=3, default height=10): {calculate_cylinder_volume(3):.2f}")
print(f"Volume (radius=3, custom height=5): {calculate_cylinder_volume(3, 5):.2f}")

# 2. Variable Positional (*args) and Keyword Arguments (**kwargs)
def train_model_report(model_name, *metrics, **hyperparameters):
    print(f"\n--- Training Report: {model_name} ---")
    print(f"Recorded Metrics: {metrics}")
    print(f"Hyperparameters: {hyperparameters}")

train_model_report("Random Forest", 0.94, 0.91, 0.92, n_estimators=100, max_depth=10, random_state=42)

# 3. Variable Scope: Local vs Global
global_counter = 100

def modify_counter():
    global global_counter
    local_val = 5
    global_counter += local_val
    print(f"Inside function: global_counter = {global_counter}")

modify_counter()
print(f"Outside function: global_counter = {global_counter}")
```

```python
# Task 20 (Continued): Lambda Expressions, Map, Filter, and Closures

# 1. Lambda Expressions
square = lambda x: x ** 2
hypotenuse = lambda a, b: (a**2 + b**2)**0.5
print(f"Lambda square(6): {square(6)}")
print(f"Lambda hypotenuse(3, 4): {hypotenuse(3, 4)}")

# 2. Map and Filter
data_points = [2, 5, 8, 11, 14, 17, 20]

# map: transform each element
scaled = list(map(lambda x: x * 1.5, data_points))
print(f"\nOriginal Data: {data_points}")
print(f"Mapped (Scaled 1.5x): {scaled}")

# filter: select elements satisfying condition
even_elements = list(filter(lambda x: x % 2 == 0, data_points))
print(f"Filtered (Evens): {even_elements}")

# 3. Inner/Nested Functions & Closures
def make_multiplier(factor):
    """Outer function returns an inner closure function."""
    def multiplier(x):
        return x * factor
    return multiplier

double = make_multiplier(2)
triple = make_multiplier(3)
print(f"\nClosure double(15): {double(15)}")
print(f"Closure triple(15): {triple(15)}")
```

## 2. Tasks 26 & 27: Exception Handling & Custom Exceptions
Demonstrating built-in exception handling structures (`try`, `except`, `else`, `finally`) and creating production-grade custom user-defined exception hierarchies.


```python
# ==========================================
# Task 26: Exception Handling in Python
# ==========================================
print("--- Task 26: Built-in Exception Handling ---")

def safe_division(numerator, denominator):
    try:
        result = numerator / denominator
    except ZeroDivisionError as e:
        print(f"[Error Handled] Division by zero attempted! Details: {e}")
        return None
    except TypeError as e:
        print(f"[Error Handled] Invalid operand type! Details: {e}")
        return None
    else:
        print(f"[Success] Computation succeeded: {numerator} / {denominator} = {result}")
        return result
    finally:
        print("[Finally Block] Cleanup executed regardless of exception state.")

# Test scenarios
print("1. Valid case:")
safe_division(20, 4)

print("\n2. Zero division case:")
safe_division(20, 0)

print("\n3. Type mismatch case:")
safe_division(20, "four")
```

```python
# ==========================================
# Task 27: User-Defined Exception Handling
# ==========================================
print("--- Task 27: User-Defined Custom Exceptions ---")

# Base custom exception for AI Training Pipeline
class MLPipelineException(Exception):
    """Base exception class for ML training workflow."""
    pass

class InvalidLearningRateError(MLPipelineException):
    """Raised when the learning rate is outside the stable range (0, 1.0]."""
    def __init__(self, lr, message="Learning rate must be in range (0, 1.0]"):
        self.lr = lr
        self.message = f"{message}. Received: {lr}"
        super().__init__(self.message)

class InsufficientSamplesError(MLPipelineException):
    """Raised when the dataset contains too few samples to train."""
    def __init__(self, sample_count, min_required=10):
        self.sample_count = sample_count
        self.message = f"Dataset has only {sample_count} samples. Minimum required is {min_required}."
        super().__init__(self.message)

def initialize_training(dataset_size, learning_rate):
    if learning_rate <= 0 or learning_rate > 1.0:
        raise InvalidLearningRateError(learning_rate)
    if dataset_size < 10:
        raise InsufficientSamplesError(dataset_size)
    print(f"Hyperparameters validated successfully: size={dataset_size}, lr={learning_rate}")

# Testing user-defined exceptions
test_cases = [
    (100, 0.01),   # Valid
    (100, 2.5),    # Invalid LR
    (5, 0.05)      # Insufficient samples
]

for samples, lr in test_cases:
    try:
        print(f"\nAttempting run with samples={samples}, lr={lr}:")
        initialize_training(samples, lr)
    except InvalidLearningRateError as e:
        print(f"Caught custom error: {e}")
    except InsufficientSamplesError as e:
        print(f"Caught custom error: {e}")
    except Exception as e:
        print(f"Caught general error: {e}")
```

## 3. Object-Oriented Programming (OOP) Suite
Comprehensive implementation of all core OOP principles required by the syllabus:

```python
# OOP 1: Classes, Instance vs Class Attributes, Constructors & Destructors
class AIModel:
    # Class Variable (Shared across all instances)
    framework = "PyTorch"
    model_count = 0

    def __init__(self, model_name, parameters_m):
        # Instance Variables (Unique to each instance)
        self.model_name = model_name
        self.parameters_m = parameters_m
        AIModel.model_count += 1
        print(f"[Constructor] Initialized '{self.model_name}' with {self.parameters_m}M parameters.")

    def summary(self):
        """Instance method accessing both instance and class attributes."""
        return f"Model: {self.model_name} | Framework: {self.framework} | Size: {self.parameters_m}M params"

    @classmethod
    def set_framework(cls, new_framework):
        """Class method modifying class-level state."""
        cls.framework = new_framework
        print(f"[ClassMethod] Updated default framework to: {new_framework}")

    @staticmethod
    def is_gpu_supported():
        """Static method independent of class or instance state."""
        return True

    def __del__(self):
        """Destructor called when object is garbage collected."""
        # Print silently or handle destruction
        pass

# Test class instantiation
bert = AIModel("BERT-Base", 110)
resnet = AIModel("ResNet-50", 25.6)
print(bert.summary())
print(resnet.summary())
print(f"Total AI models instantiated: {AIModel.model_count}")
print(f"Static GPU check: {AIModel.is_gpu_supported()}")
```

```python
# OOP 2: Inheritance Hierarchies & Method Resolution Order (MRO)

# 1. Base Class
class NeuralNetwork:
    def __init__(self, name):
        self.name = name

    def forward(self, x):
        return f"Base forward propagation for {self.name}"

# 2. Single Inheritance
class FeedForwardNN(NeuralNetwork):
    def __init__(self, name, hidden_layers):
        super().__init__(name)
        self.hidden_layers = hidden_layers

    def forward(self, x):
        return f"FeedForward ({self.name}) forward pass through {self.hidden_layers} hidden layers"

# 3. Multilevel Inheritance
class DeepMLP(FeedForwardNN):
    def __init__(self, name, hidden_layers, dropout_rate=0.2):
        super().__init__(name, hidden_layers)
        self.dropout_rate = dropout_rate

# 4. Multiple Inheritance & MRO
class FeatureExtractor:
    def extract_features(self):
        return "Features extracted"

class Classifier:
    def classify(self):
        return "Classification performed"

class VisionTransformer(FeatureExtractor, Classifier):
    def __init__(self, name):
        self.name = name

print("--- Inheritance & MRO Demonstration ---")
mlp = DeepMLP("TabularMLP", hidden_layers=4, dropout_rate=0.3)
print(mlp.forward(None))
print(f"DeepMLP dropout: {mlp.dropout_rate}")

vit = VisionTransformer("ViT-B/16")
print(f"ViT Feature Extraction: {vit.extract_features()}")
print(f"ViT Classification: {vit.classify()}")

# Inspect Method Resolution Order
print(f"VisionTransformer MRO: {[cls.__name__ for cls in VisionTransformer.mro()]}")
```

```python
# OOP 3: Access Specifiers, Name Mangling, and Property Decorators
class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder      # Public attribute
        self._branch_code = "PK-LHE-042"          # Protected attribute (convention)
        self.__balance = balance                  # Private attribute (name-mangled)

    # Property Getter
    @property
    def balance(self):
        return self.__balance

    # Property Setter with validation
    @balance.setter
    def balance(self, new_balance):
        if new_balance < 0:
            raise ValueError("Balance cannot be negative.")
        self.__balance = new_balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

print("--- Access Specifiers & Encapsulation ---")
acc = BankAccount("Hamza", 15000)
print(f"Public holder: {acc.account_holder}")
print(f"Protected branch: {acc._branch_code}")
print(f"Private balance via getter: {acc.balance}")

# Demonstrating Name Mangling
try:
    print(acc.__balance)  # Fails with AttributeError
except AttributeError as e:
    print(f"Direct private access blocked: {e}")

# Python's internal name mangling: _ClassName__attribute
print(f"Accessed via name mangling: acc._BankAccount__balance = {acc._BankAccount__balance}")

# Updating via setter
acc.balance = 20000
print(f"Updated balance: {acc.balance}")
```

```python
# OOP 4: Polymorphism, Operator Overloading, and Magic Methods
class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Operator Overloading: + operator
    def __add__(self, other):
        return Vector2D(self.x + other.x, self.y + other.y)

    # Operator Overloading: - operator
    def __sub__(self, other):
        return Vector2D(self.x - other.x, self.y - other.y)

    # Operator Overloading: * operator (Dot product or scalar multiplication)
    def __mul__(self, other):
        if isinstance(other, Vector2D):
            return (self.x * other.x) + (self.y * other.y)
        return Vector2D(self.x * other, self.y * other)

    # Equality: == operator
    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    # String representation for developers
    def __repr__(self):
        return f"Vector2D({self.x}, {self.y})"

    # String representation for users
    def __str__(self):
        return f"({self.x}i + {self.y}j)"

print("--- Operator Overloading & Dunder Methods ---")
v1 = Vector2D(3, 4)
v2 = Vector2D(1, 2)

print(f"v1 = {v1}, v2 = {v2}")
print(f"v1 + v2 = {v1 + v2}")
print(f"v1 - v2 = {v1 - v2}")
print(f"v1 * 3 (Scalar Mult) = {v1 * 3}")
print(f"v1 * v2 (Dot Product) = {v1 * v2}")
print(f"v1 == Vector2D(3, 4): {v1 == Vector2D(3, 4)}")
```

```python
# OOP 5: Abstract Base Classes (ABC) and Modern Data Classes
from abc import ABC, abstractmethod
from dataclasses import dataclass

# Abstract Base Class
class LossFunction(ABC):
    @abstractmethod
    def compute(self, y_true, y_pred):
        """Must be overridden by all concrete loss functions."""
        pass

class MeanSquaredError(LossFunction):
    def compute(self, y_true, y_pred):
        return sum((yt - yp)**2 for yt, yp in zip(y_true, y_pred)) / len(y_true)

class MeanAbsoluteError(LossFunction):
    def compute(self, y_true, y_pred):
        return sum(abs(yt - yp) for yt, yp in zip(y_true, y_pred)) / len(y_true)

# Modern Python Data Class
@dataclass
class DatasetConfig:
    name: str
    num_samples: int
    num_features: int
    train_split: float = 0.8

print("--- Abstract Classes and Data Classes ---")
mse = MeanSquaredError()
mae = MeanAbsoluteError()

y_t = [3.0, 5.0, 2.5]
y_p = [2.8, 5.2, 2.0]
print(f"MSE Loss: {mse.compute(y_t, y_p):.4f}")
print(f"MAE Loss: {mae.compute(y_t, y_p):.4f}")

cfg = DatasetConfig(name="Housing Prices", num_samples=5000, num_features=12)
print(f"DataClass Config: {cfg}")
```

## 4. Descriptive Statistics & Probability: Tasks 49 & 50
Complete mathematical implementations and analysis of:
- **Data Types**: Structured, Unstructured, Quantitative (Continuous, Discrete), Qualitative (Nominal, Ordinal, Binary)
- **Task 49**: Measures of Central Tendency (Mean, Median, Mode) from scratch and with built-in/SciPy functions
- **Task 50**: Measures of Dispersion (Variance, Standard Deviation, Range, IQR, Coefficient of Variation, Skewness, Kurtosis)
- **Measures of Position**: Z-Scores, Percentiles, and Quartiles


```python
# ==========================================
# Task 49: Perform Mean, Median, and Mode
# ==========================================
import statistics
import numpy as np

dataset = [12, 15, 12, 18, 22, 25, 12, 30, 18, 22, 19, 21, 24]
print(f"Dataset (N={len(dataset)}): {dataset}")

# 1. Implementation from Scratch
# Mean: Sum of elements divided by total count
mean_scratch = sum(dataset) / len(dataset)

# Median: Middle value of sorted sequence
sorted_data = sorted(dataset)
n = len(sorted_data)
if n % 2 == 1:
    median_scratch = sorted_data[n // 2]
else:
    median_scratch = (sorted_data[n // 2 - 1] + sorted_data[n // 2]) / 2

# Mode: Most frequently occurring value
freq = {}
for val in dataset:
    freq[val] = freq.get(val, 0) + 1
mode_scratch = max(freq, key=freq.get)

print("\n--- Task 49: Central Tendency (From Scratch) ---")
print(f"Mean:   {mean_scratch:.4f}")
print(f"Median: {median_scratch:.4f}")
print(f"Mode:   {mode_scratch} (Frequency: {freq[mode_scratch]})")

# 2. Verification with statistics and numpy
print("\n--- Verification with Python Libraries ---")
print(f"statistics.mean:   {statistics.mean(dataset):.4f}")
print(f"statistics.median: {statistics.median(dataset):.4f}")
print(f"statistics.mode:   {statistics.mode(dataset)}")
```

```python
# ==========================================
# Task 50: Perform Standard Deviation & Dispersion
# ==========================================
import math
import numpy as np
from scipy import stats

# 1. Variance and Standard Deviation from Scratch
n = len(dataset)
mean_val = sum(dataset) / n

# Population vs Sample Variance
sample_variance = sum((x - mean_val) ** 2 for x in dataset) / (n - 1)
sample_std_dev = math.sqrt(sample_variance)

population_variance = sum((x - mean_val) ** 2 for x in dataset) / n
population_std_dev = math.sqrt(population_variance)

# 2. Range and Interquartile Range (IQR)
data_range = max(dataset) - min(dataset)
q25, q50, q75 = np.percentile(dataset, [25, 50, 75])
iqr = q75 - q25

# 3. Coefficient of Variation (CV = (std / mean) * 100)
cv = (sample_std_dev / mean_val) * 100

# 4. Skewness and Kurtosis
skew_val = stats.skew(dataset)
kurt_val = stats.kurtosis(dataset)

print("--- Task 50: Measures of Dispersion & Shape ---")
print(f"Sample Variance:         {sample_variance:.4f}")
print(f"Sample Standard Dev:     {sample_std_dev:.4f}")
print(f"Population Std Dev:      {population_std_dev:.4f}")
print(f"Data Range:              {data_range}")
print(f"Quartiles:               Q1(25%)={q25:.2f}, Q2(Median)={q50:.2f}, Q3(75%)={q75:.2f}")
print(f"Interquartile Range IQR: {iqr:.2f}")
print(f"Coefficient of Var (CV): {cv:.2f}%")
print(f"Skewness:                {skew_val:.4f} ({'Right-skewed' if skew_val > 0 else 'Left-skewed'})")
print(f"Kurtosis:                {kurt_val:.4f}")
```

```python
# Measures of Position: Z-Scores Calculation and Outlier Detection
# Z-Score formula: z = (x - mean) / std

z_scores = [(x - mean_val) / sample_std_dev for x in dataset]

print("--- Measures of Position: Z-Scores ---")
print(f"{'Value':<10} {'Z-Score':<15} {'Status':<15}")
print("-" * 40)
for val, z in zip(dataset, z_scores):
    is_outlier = abs(z) > 2.0
    status = "Potential Outlier" if is_outlier else "Normal"
    print(f"{val:<10} {z:<15.4f} {status:<15}")

# Percentile lookup function
def calculate_percentile(data, target_value):
    count_below = sum(1 for x in data if x <= target_value)
    return (count_below / len(data)) * 100

query_val = 22
print(f"\nPercentile rank of value {query_val}: {calculate_percentile(dataset, query_val):.1f}%")
```
